```python
import numpy as np
import pandas as pd
import torch
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import LabelEncoder
import torch.nn as nn
```

```python
df = pd.read_csv('https://raw.githubusercontent.com/gscdit/Breast-Cancer-Detection/refs/heads/master/data.csv')
df.head()
```

```python
df.shape
```

```python
df.drop(columns=['id', 'Unnamed: 32'], inplace=True)
df.head()
```

## Train Test Split

```python
x_train, x_test, y_train, y_test = train_test_split(df.iloc[:, 1:], df.iloc[:, 0], test_size=0.2)
```

```python
scaler = StandardScaler()
x_train = scaler.fit_transform(x_train)
x_test = scaler.transform(x_test)
```

```python
x_train
```

```python
encoder = LabelEncoder()
y_train = encoder.fit_transform(y_train)
y_test = encoder.transform(y_test)
```

```python
y_train
```

```python
x_train_tensor = torch.from_numpy(x_train).float()
x_test_tensor = torch.from_numpy(x_test).float()
y_train_tensor = torch.from_numpy(y_train).float()
y_test_tensor = torch.from_numpy(y_test).float()
```

```python
x_train_tensor.shape
```

```python
from torch.utils.data import  Dataset, DataLoader
class CustomDataset(Dataset):
  def __init__(self, features, lables):
    self.features = features
    self.lables = lables

  def __len__(self):
    return len(self.features)

  def __getitem__(self, idx):
    return self.features[idx], self.lables[idx]
```

```python
train_dataset = CustomDataset(x_train_tensor, y_train_tensor)
test_dataset = CustomDataset(x_test_tensor, y_test_tensor)
```

```python
train_dataset[10]
```

```python
train_loader = DataLoader(train_dataset, batch_size=32, shuffle=True)
test_loader = DataLoader(test_dataset, batch_size=32, shuffle=False)
```

## Defining the Model

```python
class MySimpleNN(nn.Module):
  def __init__(self, num_features):
    super().__init__()
    self.linear = nn.Linear(num_features, 1)
    self.sigmoid = nn.Sigmoid()

  def forward(self, features):
    out = self.linear(features)
    out = self.sigmoid(out)
    return out
```

### Important Paramenters

```python
learning_rate = 0.1
epochs = 25
```

```python
loss_function = nn.BCELoss()
```

```python
type(loss_function)
```

### Training Pipeline

```python
# Create Model
model = MySimpleNN(x_train_tensor.shape[1])
optimizer = torch.optim.SGD(model.parameters(), lr=learning_rate)
# Define Loop
for epoch in range(epochs):
  for batch_features, batch_lables in train_loader:
    # Forward Pass
    y_pred = model(batch_features)
    # Loss Calculate
    loss = loss_function(y_pred, batch_lables.view(-1, 1))
    # Zero Gradients
    optimizer.zero_grad()
    # Backward class
    loss.backward()
    # Parameters Update
    optimizer.step()
    # Print Loss in each Epoch
  print(f'Epoch: {epoch+1}, Loss: {loss.item()}')
```

### Evaluation

```python
# Model Evaluation using test_loader
model.eval()
accuracy_list = []

with torch.no_grad():
  for batch_features, batch_lables in test_loader:
    y_pred = model(batch_features)
    y_pred = (y_pred > 0.8).float()
    batch_accuracy = (y_pred.view(-1) == batch_lables).float().mean().item()
    accuracy_list.append(batch_accuracy)

overall_accuracy = sum(accuracy_list) / len(accuracy_list)
print(f"Accuracy: {overall_accuracy:.4f}")
# print(y_pred)
```
