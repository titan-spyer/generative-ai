## Reduction Operation

```python
e = torch.randint(size=(2, 3), low=0, high=10, dtype=torch.float32)
```

```python
# sum
torch.sum(e)
# sum along column
torch.sum(e, dim=0)
# Sum Along rows
torch.sum(e, dim=1)
```

```python
# Mean
torch.mean(e)
# Mean along column
torch.mean(e, dim=0)
# Mean along rows
torch.mean(e, dim=1)
```

```python
# Median
torch.median(e)
# Median along column
torch.median(e, dim=0)
# Median along rows
torch.median(e, dim=1)
```

```python
# Max and min
torch.max(e)
torch.min(e)
```

```python
# Product
torch.prod(e)
```

```python
# standard Deviation
torch.std(e)
```

```python
# Variance
torch.var(e)
```

```python
torch.argmax(e)
```

```python
torch.argmin(e)
```

### Matrix operation

```python
c= torch.randint(size=(2, 3), low=0, high=10, dtype=torch.float32)
d= torch.randint(size=(3, 2), low=0, high=10, dtype=torch.float32)
```

```python
# Matrix multipication
torch.matmul(c, d)
```

```python
vector1 = torch.tensor([1, 2])
vector2 = torch.tensor([3, 4])

# Dot Product
torch.dot(vector1, vector2)
```

```python
# Transpose
torch.transpose(c, 0, 1)
```

```python
# Determinant
# The torch.det() function requires a square matrix (number of rows equals number of columns).
# The tensor 'e' is a 2x3 matrix, which is not square, causing the RuntimeError.
# Let's create a square matrix to demonstrate its usage.
square_matrix = torch.tensor([[1, 2], [3, 4]], dtype=torch.float32)
print(f"Square Matrix:\n{square_matrix}")
print(f"Determinant of square_matrix: {torch.det(square_matrix)}")
```

```python
torch.inverse(square_matrix)
```

### Comparision Operation

```python
i = torch.randint(size=(2, 3), low=0, high=10, dtype=torch.float32)
j = torch.randint(size=(2, 3), low=0, high=10, dtype=torch.float32)
```

```python
# Greater than
i > j
# Less Than
i < j
# Equal to
i == j
# Not equal to
i != j
```
